# SHOPIFY STOCK – DATA ANALYSIS REPORT

## 1. Introduction

Shopify Stock Data Analysis is a project that focuses on studying the historical stock market performance of Shopify using Python. Stock market data contains important information such as opening price, closing price, highest price, lowest price and trading volume. By analyzing this data, we can understand stock price movements, market trends and changes in trading activities.

This project is divided into two parts: Time-Series Preparation and Financial Visualization.

## 2. Objective

The main objective of this project is to analyze Shopify stock market data and understand its historical performance. It focuses on preparing time-series data, calculating daily returns, tracking trading volume, identifying price fluctuations and visualizing financial trends.

## 3. Time-Series Preparation

Time-Series Preparation is used to organize and analyze stock market data based on dates.

The following operations were performed:

- **Date Formatting:** Converted the date column into datetime format to perform time-based analysis.
- **Daily Price Changes:** Calculated the difference between the current day's closing price and the previous day's closing price.
- **Daily Returns:** Calculated daily percentage returns to understand the increase or decrease in stock prices.
- **Volume Tracking:** Analyzed daily trading volume to understand the number of shares traded.
- **Price Fluctuation Bounds:** Calculated the difference between the highest and lowest stock prices to identify daily price variations.

These operations helped prepare the dataset for further financial analysis.

## 4. Financial Visualization

Financial Visualization helps understand stock market performance through graphs and charts.

The following analyses were performed:

- **OHLC Visual Trends:** Visualized Open, High, Low and Close stock prices to understand daily market movements.
- **Moving Averages:** Calculated 20-day and 50-day moving averages to identify short-term and long-term price trends.
- **Risk Analysis:** Analyzed daily returns to understand changes in stock performance.
- **Volatility Metrics:** Calculated standard deviation to measure fluctuations in stock returns.
- **Return Correlation Heatmap:** Created a correlation heatmap to identify relationships between different stock variables.
- **Daily Returns Visualization:** Used line graphs to observe daily percentage changes in stock prices.

## 5. Tools and Technologies Used

1. Python – Programming language used for data analysis.
2. Pandas – Used for data cleaning and manipulation.
3. NumPy – Used for numerical calculations.
4. Matplotlib – Used for creating graphs and visualizations.
5. Seaborn – Used for creating correlation heatmaps.
6. Jupyter Notebook – Used to execute Python programs.

## 6. Key Findings

- Historical stock prices helped identify changes in Shopify's market performance.
- Daily returns showed the percentage changes in closing prices.
- Trading volume provided information about daily market activity.
- Moving averages helped identify stock price trends.
- Volatility analysis measured the level of variation in stock returns.
- Correlation analysis helped understand relationships between stock variables.

## 7. Conclusion

The Shopify Stock Data Analysis project was successfully performed using Python libraries such as Pandas, NumPy, Matplotlib and Seaborn. Time-series preparation helped organize stock market data and calculate important financial values. Financial visualization helped identify stock price trends, moving averages, daily returns and volatility.
