```python
# Text Cleaning - Remove Extra Spaces
df["Gender"] = df["Gender"].str.strip()
df["Medical Condition"] = df["Medical Condition"].str.strip()
df["Admission Type"] = df["Admission Type"].str.strip()
df["Insurance Provider"] = df["Insurance Provider"].str.strip()   
```

```python
# Text Standardization
df["Gender"] = df["Gender"].str.title()
df["Medical Condition"] = df["Medical Condition"].str.title()
df["Admission Type"] = df["Admission Type"].str.title()
df["Insurance Provider"] = df["Insurance Provider"].str.title()  
```

```python
# Verify Cleaned Text Values
print(df["Gender"].unique())
print(df["Medical Condition"].unique())
print(df["Admission Type"].unique())
print(df["Insurance Provider"].unique())
```
